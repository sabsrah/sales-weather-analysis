## 4. Convert Data Types

```python
# Convert the date columns from object to datetime
dfi['Date'] = pd.to_datetime(dfi['Date'])
dfi.info()
dfw['WeekStarting'] = pd.to_datetime(dfw['WeekStarting'])
dfw.info()
```

## 6. Identify and Handle Missing Values, Defend reasoning

```python
dfi.isna().sum()
```

```python
# Remove rows missing Price or TotalDue
dfi = dfi.dropna(subset=['Price', 'TotalDue']).reset_index(drop=True)

# Verify that the missing values were removed by printing count of missing values again
dfi.isna().sum()
```

The 12 rows with missing Price and TotalDue values were removed because they represented less than 0.003% of the dataset. Imputing both values could introduce inaccurate sales amounts.


## 7. Aggregate dfi at the Daily Level, display first 10 rows, defend reasoning

```python
dfid = (
    dfi.groupby('Date')[['Quantity', 'TotalDue']]
       .sum()
       .reset_index()
)

dfid.head(10)
```

Sum was used for quantity to calculate the total quantity recorded each day, and for TotalDue to calculate daily net sales including the effect of returns or refunds. Sum was chosen as the aggregate measures since the total quantity and total sales amount for each day is of interest to the analysis.

## 8. Create dfwd by converting dfw from wide to long format, display first 10 rows

```python
dfwd = dfw.melt(
    id_vars=['Week', 'WeekStarting'],
    value_vars=[
        'Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday'
    ],
    var_name='day',
    value_name='max_temp'
)
dfwd.head(10)
```

## 9. Create and Sort the Date Column in dfwd

```python
# Number of days after WeekStarting
day_offsets = {
    'Sunday': 0,
    'Monday': 1,
    'Tuesday': 2,
    'Wednesday': 3,
    'Thursday': 4,
    'Friday': 5,
    'Saturday': 6
}

# Create the Date column
dfwd['Date'] = (
    dfwd['WeekStarting']
    + pd.to_timedelta(dfwd['day'].map(day_offsets), unit='D')
)

# Sort the DataFrame by Date
dfwd = dfwd.sort_values('Date').reset_index(drop=True)

# Display the first 10 rows
dfwd.head(10)
```

We know that the Date column was created correctly because each weekday is observed to be matched with the appropriate date after WeekStarting. For example, the week beginning Sunday, January 3, 2016, shows Monday as January 4, 2016, and the results are arranged in ascending date order.

## 10. Join the DataFrames

```python
df = pd.merge(
    dfid,
    dfwd,
    on='Date',
    how='inner'
)

df = df.sort_values('Date').reset_index(drop=True)
print(dfid.shape, dfwd.shape, df.shape)
df.head(10)
```

*dfid has 1,021 rows (it covers 2014 through 2017) but dfwd only has 364 rows (just 2016), so an inner join should cut things down to just the 2016 dates that show up in both. That's what happened: df ends up with 311 rows, and every row has both Quantity/TotalDue and max_temp filled in, no blanks. That's the join working as expected.*

## 11. Create and Analyze Scatter Plots

```python


df.plot.scatter(
    x='max_temp',
    y='TotalDue',
    title='Maximum Temperature vs. Total Due',
    figsize=(8, 5),
    alpha=1,
    color='steelblue',
    edgecolor='black',
    linewidth=0.3,
    s=40
)
plt.xlabel('Max Temperature (°F)')
plt.ylabel('Total Due ($)')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
df.plot.scatter(
    x='max_temp',
    y='Quantity',
    title='Maximum Temperature vs. Quantity',
    figsize=(8, 5),
    alpha=1,
    color='steelblue',
    edgecolor='black',
    linewidth=0.3,
    s=40
)
plt.xlabel('Max Temperature (°F)')
plt.ylabel('Quantity')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
df.plot.scatter(
    x='Quantity',
    y='TotalDue',
    title='Quantity vs. Total Due',
    figsize=(8, 5),
    alpha=1,
    color='steelblue',
    edgecolor='black',
    linewidth=0.3,
    s=40
)
plt.xlabel('Quantity')
plt.ylabel('Total Due ($)')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
